```python
import  json

 # Function to load JSON data from a file
def load_data(filename):
    with open(filename , "r") as f:
        return json.load(f)
        

 # Function to find pages a user might like based on common interests
def pages_you_might_like(data , user_id):
    # Dictionary containing all users id and their liked pages
    user_pages = {}
    for user in data["users"]:
        user_pages[user['id']] = set(user['liked_pages'])

     # If the user is not found, return an empty list
    if user_id not in user_pages:
        return []

    # Liked pages of current user
    user_liked_pages = user_pages[user_id]

    page_suggestions = {}

    # Other users and their liked pages
    for other_user, pages in user_pages.items():
        if other_user != user_id:
            similar_pages = user_liked_pages.intersection(pages)    # Similar liked pages of Current and other user    

            # loop on Page in all pages liked by other user
            for page in pages:
                if page not in user_liked_pages:
            
                    page_suggestions[page] = page_suggestions.get(page , 0) + len(similar_pages)
    
    # Sorting of Dict. based on no. of similar pages in Descending order
    sorted_pages = sorted(page_suggestions.items() , key = lambda x:x[1] , reverse = True)

    return [page_id for page_id , score in sorted_pages]
    


data = load_data("massive_data.json")
user_id = 15

pages_recommendation = pages_you_might_like(data , user_id)
print(f"Pages you might like for user {user_id} : {pages_recommendation}")
```
